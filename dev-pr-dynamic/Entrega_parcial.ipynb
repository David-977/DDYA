## 1) Descripción del problema **[1 pts]**


Un vivero está analizando el crecimiento de la producción de nuevas plantas a partir de un
proceso de reproducción controlada.
Se ha observado que el número de nuevas plantas producidas en el mes n sigue el siguiente
comportamiento:
• En el mes 0 se produjeron 0 nuevas plantas.
• En el mes 1 se produjo 1 nueva planta.
• A partir del mes 2, el número de nuevas plantas producidas es igual a la suma de las plantas
producidas durante los dos meses anteriores, debido a que las plantas que alcanzan su etapa de
reproducción generan nuevos brotes siguiendo este patrón de crecimiento.
Relación de recurrencia:
T(n) = 0 si n = 0
T(n) = 1 si n = 1
T(n) = T(n-1) + T(n-2) si n ≥ 2
Determinar con memoización cuántas nuevas plantas serán producidas en el mes n.

Justificacion:
El problema se parece a una sucesión de fibonacci mas o menos, ya que la cantidad de plantas producidas en el mes n se obtiene sumando las cantidades producidas en los dos meses anteriores. Por esta razón, la solución recursiva tiene que resolver el mismo problema varias veces.
 
Para evitar estos cálculos repetidos se utiliza memoización, almacenando los resultados que ya fueron calculados. Así, cada mes se procesa una sola vez.

Funcion T(n):
T(n) es lineal: T(n)=O(n)


## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
def pedir_mes():
    while True:
        try:
            mes=int(input("ingrese el numero del mes que sedea consultar"))

            if mes>=0:
                return mes
            else:
                print("El mes no puede ser negativo")

        except ValueError:
            print("Debe ingresar un numero entero")

def calcular_produccion(mes, memo):
    if mes in memo:
        return memo[mes]
    if mes==0:
        return 0
    if mes==1:
        return 1

    produccion=(calcular_produccion(mes-1,memo)+calcular_produccion(mes-2,memo))

    memo[mes]=produccion
    return produccion 

def main():
    mes=pedir_mes()
    memo={}
    resultado=calcular_produccion(mes,memo)

    print()
    print("Mes consultado:",mes)
    print("Cantidad de nuevas plantas producidas:",resultado)

main()
```
