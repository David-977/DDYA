### 5) Análisis de complejidad **[15 pts]**
Incluye:
- Justificación
- Función T[n]
- Mejor y peor caso con gráfica


- Justificación
Analizamos BFS porque es el algoritmo que mas trabajo hace, y la busqueda de ruta usa BFS por dentro. DFS sigue el mismo patron. Sea n el numero de usuarios y E el numero de amistades. BFS hace tres cosas:
Prepara la cola y el registro de visitados, lo que cuesta un paso fijo.
 
Saca de la cola a cada usuario una sola vez, porque el registro de visitados evita repetirlos. Eso suma n pasos.
Revisa la lista de amigos de cada usuario que saca. Como cada amistad esta guardada en las dos listas, en total se revisan 2E amigos.
 
- Función T[n]
 
T(n) = 1 + n + 2E
 
- Mejor y peor caso con gráfica
 
- Mejor caso
El usuario inicial no tiene amigos. BFS lo saca de la cola, no encuentra a nadie mas y termina, sin importar el tamaño de la red. Aqui n y E no influyen:
O(1)
- Peor caso
BFS tiene que recorrer toda la red. Pasa, por ejemplo, cuando el destino de una ruta está muy lejos o no tiene conexión con el origen. Entonces:
 O(n + E)

## 6) Código funcional **[15 pts]**

```python
import matplotlib.pyplot as plt
import math
 
 
grafo = {
    "Ana": ["Carlos", "Laura"],
    "Carlos": ["Ana", "Mateo", "Sofia"],
    "Laura": ["Ana", "Daniel"],
    "Mateo": ["Carlos"],
    "Sofia": ["Carlos", "Valentina"],
    "Daniel": ["Laura", "Valentina"],
    "Valentina": ["Sofia", "Daniel"]
}
 
 
def pedir_entero(mensaje):
    while True:
        try:
            numero = int(input(mensaje))
            return numero
        except ValueError:
            print("Debe ingresar un número entero.")
 
 
def pedirred(grafo, mensaje):
    red = input(mensaje).strip().title()
 
    while red not in grafo:
        print("La red no existe.")
        red = input(mensaje).strip().title()
 
    return red
 
 
def mostrar_grafo(grafo):
    print("LISTA DE ADYACENCIA")
 
    for red in grafo:
        print(red, "->", grafo[red])
 
 
def mostrar_grafico(grafo):
    posiciones = {}
    redes = list(grafo)
    cantidad = len(redes)
 
    for rep in range(cantidad):
        angulo = 2 * math.pi * rep / cantidad
        x = 3 * math.cos(angulo)
        y = 3 * math.sin(angulo)
        posiciones[redes[rep]] = (x, y)
 
    plt.figure(figsize=(9, 7))
 
    conexiones = []
 
    for red in grafo:
        for vecino in grafo[red]:
            if (vecino, red) not in conexiones:
                x1, y1 = posiciones[red]
                x2, y2 = posiciones[vecino]
 
                plt.plot([x1, x2], [y1, y2],
                         color="black", zorder=1)
 
                conexiones.append((red, vecino))
 
    for red in grafo:
        x, y = posiciones[red]
 
        plt.scatter(x, y, s=1800, color="white",
                    edgecolors="black", zorder=2)
 
        plt.text(x, y, red, fontsize=10,
                 ha="center", va="center", zorder=3)
 
    plt.title("Red Social")
    plt.axis("off")
    plt.show()
 
 
def bfs_niveles(grafo, inicio):
    visitados = []
    cola = []
    niveles = {}
 
    cola.append(inicio)
    visitados.append(inicio)
    niveles[inicio] = 0
 
    while len(cola) > 0:
        actual = cola.pop(0)
 
        for vecino in grafo[actual]:
            if vecino not in visitados:
                visitados.append(vecino)
                cola.append(vecino)
                niveles[vecino] = niveles[actual] + 1
 
    return visitados, niveles
 
 
def dfs(grafo, red, visitados):
    visitados.append(red)
 
    for vecino in grafo[red]:
        if vecino not in visitados:
            dfs(grafo, vecino, visitados)
 
 
def buscar_bfs(grafo, inicio, destino):
    cola = []
    visitados = []
    padres = {}
 
    cola.append(inicio)
    visitados.append(inicio)
    padres[inicio] = None
 
    while len(cola) > 0:
        actual = cola.pop(0)
 
        if actual == destino:
            conexion = []
 
            while actual is not None:
                conexion.append(actual)
                actual = padres[actual]
 
            conexion.reverse()
            return conexion
 
        for vecino in grafo[actual]:
            if vecino not in visitados:
                visitados.append(vecino)
                cola.append(vecino)
                padres[vecino] = actual
 
    return None
 
 
def mostrar_bfs(grafo, inicio):
    recorrido, niveles = bfs_niveles(grafo, inicio)
 
    print("Recorrido BFS:")
    print("->".join(recorrido))
 
    print("Niveles:")
 
    for nivel in range(max(niveles.values()) + 1):
        redes_nivel = []
 
        for red in recorrido:
            if niveles[red] == nivel:
                redes_nivel.append(red)
 
        print("Nivel", nivel, ":", ", ".join(redes_nivel))
 
 
def mostrar_dfs(grafo, inicio):
    visitados = []
 
    dfs(grafo, inicio, visitados)
 
    print("Recorrido DFS:")
    print(" -> ".join(visitados))
 
 
def buscar_conexion(grafo):
    inicio = pedirred(grafo, "red de origen: ")
    destino = pedirred(grafo, "red de destino: ")
 
    conexion = buscar_bfs(grafo, inicio, destino)
 
    if conexion is not None:
        print("conexion encontrada:")
        print(" -> ".join(conexion))
        print("Cantidad de conexiones:", len(conexion) - 1)
    else:
        print("No existe una conexion entre las personas.")
 
 
def agregar_red(grafo):
    nueva = input("Ingrese la nueva red: ").strip().title()
 
    while nueva == "":
        print("El nombre no puede estar vacío.")
        nueva = input("Ingrese la nueva red: ").strip().title()
 
    if nueva in grafo:
        print("La red ya existe.")
        return False
 
    grafo[nueva] = []
 
    while len(grafo[nueva]) < 2:
        conexion = input(
            "Conectar con una red existente (FIN para terminar): "
        ).strip().title()
 
        if conexion == "Fin":
            if len(grafo[nueva]) >= 2:
                break
            print("Debe registrar al menos dos conexiones.")
            continue
 
        if conexion == nueva:
            print("Una red no puede conectarse consigo misma.")
 
        elif conexion not in grafo:
            print("La red indicada no existe.")
 
        elif conexion in grafo[nueva]:
            print("Esa conexión ya fue registrada.")
 
        else:
            grafo[nueva].append(conexion)
            grafo[conexion].append(nueva)
            print("Conexión agregada.")
 
    print("red agregada correctamente.")
    return True
 
 
def menu():
    print("------ REDES ------")
    print("1. Mostrar grafo")
    print("2. Recorrido BFS desde Ana")
    print("3. Recorrido DFS desde Ana")
    print("4. Seleccionar red inicial")
    print("5. Buscar conexion entre personas")
    print("6. Agregar red")
    print("7. Salir")
 
 
def main():
    opcion = 0
 
    while opcion != 7:
        menu()
        opcion = pedir_entero("Seleccione una opción: ")
 
        while opcion < 1 or opcion > 7:
            print("Opción inválida.")
            opcion = pedir_entero("Seleccione una opción: ")
 
        if opcion == 1:
            mostrar_grafo(grafo)
            mostrar_grafico(grafo)
 
        elif opcion == 2:
            mostrar_bfs(grafo, "Ana")
 
        elif opcion == 3:
            mostrar_dfs(grafo, "Ana")
 
        elif opcion == 4:
            inicio = pedirred(
                grafo, "Ingrese la red inicial: "
            )
 
            mostrar_bfs(grafo, inicio)
            mostrar_dfs(grafo, inicio)
 
        elif opcion == 5:
            buscar_conexion(grafo)
 
        elif opcion == 6:
            agregada = agregar_red(grafo)
 
            if agregada:
                mostrar_grafo(grafo)
                mostrar_grafico(grafo)
 
                inicio = pedirred(
                    grafo, "red inicial para los recorridos: "
                )
 
                mostrar_bfs(grafo, inicio)
                mostrar_dfs(grafo, inicio)
 
                buscar = input(
                    "¿Desea buscar una conexion? (SI/NO): "
                ).strip().upper()
 
                if buscar == "SI":
                    buscar_conexion(grafo)
 
        elif opcion == 7:
            print("Finalizando programa...")
 
 
main()
```
